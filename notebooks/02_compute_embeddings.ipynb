```python
import os, sys
sys.path.append(os.path.abspath(".."))

import pandas as pd
import numpy as np
from src.embeddings import embed_lyrics, save_embeddings
```

```python
# Cell 2: Configuration
SAMPLE_PATH = "../data/sample_stratified.csv"
EMBED_OUT  = "../results/embeddings_stratified.npy"
META_OUT   = "../results/metadata_stratified.csv"
MODEL_NAME = "all-mpnet-base-v2"
BATCH_SIZE = 16        # smaller to reduce memory use
DEVICE = "mps"         # or "cuda" if you have GPU
df = pd.read_csv(SAMPLE_PATH)
lyrics_list = df["clean_lyrics"].tolist()
# Cell 4: Compute Embeddings
embeddings = embed_lyrics(
    lyrics_list,
    model_name=MODEL_NAME,
    batch_size=BATCH_SIZE,
    device=DEVICE,
)
```

```python
save_embeddings(embeddings, EMBED_OUT)
# save tags and any other metadata alongside indices
df[["tag"]].to_csv(META_OUT, index=False)
print("Saved embeddings to", EMBED_OUT)
print("Saved metadata to", META_OUT)
```
